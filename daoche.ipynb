```python
from ultralytics import YOLO

# Load a pretrained YOLO26n model
model = YOLO("yolo26n.pt")
```

```python
# Train the model on the NEU-DET dataset for 200 epochs
results = model.train(
    data="datasets/NEU-DET/neu-det.yaml",  # Path to dataset configuration file
    epochs=200,  # Number of training epochs
    imgsz=640,  # Image size for training
    device=0,  # Device to run on (e.g., 'cpu', 0, [0,1,2,3])
    batch=-1, cache='ram', optimizer='SGD', patience=20, workers=8,
    augment=True, box=7.5, close_mosaic=20, cls=0.5, cos_lr=True, lr0=0.01, overlap_mask=True,
)
```

```python
# Evaluate the model's performance on the test set
metrics = model.val(split='test')
```
